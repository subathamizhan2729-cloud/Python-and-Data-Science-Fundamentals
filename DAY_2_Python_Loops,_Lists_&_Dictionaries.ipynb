```python

# DAY 2 - Python Loops, Lists & Dictionaries

# 1. For Loop - Print 1 to 10
print("Numbers 1 to 10:")
for i in range(1, 11):
    print(i)

# 2. Even Numbers
print("\nEven Numbers:")
for i in range(1, 51):
    if i % 2 == 0:
        print(i)

# 3. Odd Numbers
print("\nOdd Numbers:")
for i in range(1, 51):
    if i % 2 != 0:
        print(i)

# 4. Multiplication Table
num = int(input("\nEnter a number for multiplication table: "))

print("\nMultiplication Table:")
for i in range(1, 11):
    print(num, "x", i, "=", num * i)

# 5. Sum of 1 to 10
total = 0

for i in range(1, 11):
    total += i

print("\nSum of 1 to 10:", total)

# 6. List - Student Marks
marks = [80, 75, 90, 65, 88]

print("\nStudent Marks:", marks)
print("Highest Mark:", max(marks))
print("Lowest Mark:", min(marks))
print("Total Mark:", sum(marks))
print("Average Mark:", sum(marks) / len(marks))

# 7. Dictionary - Student Details
student = {
    "name": "Hariprasath",
    "age": 20,
    "course": "AI & DS",
    "college": "K.S.K College of Engineering and Technology"
}

print("\nStudent Details:")
print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("College:", student["college"])
```

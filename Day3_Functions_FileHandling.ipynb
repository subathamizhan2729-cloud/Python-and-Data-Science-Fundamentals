```python
# ==========================================
# DAY 3 - FUNCTIONS & FILE HANDLING
# ==========================================

# 1. Addition Function
def addition(a, b):
    return a + b

print("Addition:", addition(10, 20))


# 2. Subtraction Function
def subtraction(a, b):
    return a - b

print("Subtraction:", subtraction(20, 10))


# 3. Multiplication Function
def multiplication(a, b):
    return a * b

print("Multiplication:", multiplication(10, 5))


# 4. Division Function
def division(a, b):
    if b != 0:
        return a / b
    else:
        return "Cannot divide by zero"

print("Division:", division(20, 5))


# 5. Student Details Function
def student_details(name, age, course):
    print("\nStudent Details")
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student_details("Hariprasath", 20, "AI & DS")


# ==========================================
# FILE HANDLING
# ==========================================

# 6. Create and Write a File
file = open("student.txt", "w")

file.write("Name: Hariprasath\n")
file.write("Age: 20\n")
file.write("Course: AI & Data Science\n")

file.close()

print("\nFile created and data written successfully!")


# 7. Read the File
file = open("student.txt", "r")

data = file.read()

print("\nFile Content:")
print(data)

file.close()


# 8. Append Data
file = open("student.txt", "a")

file.write("Module: Python & Data Science Fundamentals\n")

file.close()

print("New data added successfully!")


# 9. Read Updated File
file = open("student.txt", "r")

print("\nUpdated File Content:")
print(file.read())

file.close()
```

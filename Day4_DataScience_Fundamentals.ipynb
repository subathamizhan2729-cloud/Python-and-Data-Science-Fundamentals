```python

# ==========================================
# DAY 4 - DATA SCIENCE FUNDAMENTALS
# ==========================================

# 1. NumPy
import numpy as np

marks = np.array([80, 75, 90, 65, 88])

print("Marks:", marks)
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))


# ==========================================
# 2. Pandas
# ==========================================

import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Kumar", "Ravi", "Siva"],
    "Python": [80, 75, 90, 65, 88],
    "Maths": [85, 70, 95, 60, 82]
}

df = pd.DataFrame(data)
8
print("\nStudent Data:")
print(df)

print("\nData Information:")
print(df.info())

print("\nPython Average:", df["Python"].mean())
print("Maths Average:", df["Maths"].mean())

print("\nHighest Python Mark:")
print(df["Python"].max())


# ==========================================
# 3. Data Analysis
# ==========================================

df["Total"] = df["Python"] + df["Maths"]

df["Average"] = df["Total"] / 2

print("\nUpdated Student Data:")
print(df)


# ==========================================
# 4. Find Top Student
# ==========================================

top_student = df.loc[df["Total"].idxmax()]

print("\nTop Student:")
print(top_student)


# ==========================================
# 5. Data Visualization
# ==========================================

import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Python"])

plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.title("Student Python Marks")

plt.show()
```
